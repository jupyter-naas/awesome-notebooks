```python
import requests
import pandas as pd
```

```python
# The name of the funder to search for
search_term = "Gates Foundation"

# The base URL for the OpenAlex funders endpoint
url = "https://api.openalex.org/funders"
```

```python
def search_funders(query, base_url):
    """
    Searches for funders in OpenAlex based on a display name.

    Args:
        query (str): The name of the funder to search for.
        base_url (str): The base URL for the OpenAlex API.

    Returns:
        dict: The JSON response from the API, or None if there's an error.
    """
    params = {
        'search': query
    }
    try:
        response = requests.get(base_url, params=params)
        response.raise_for_status()  # This will raise an HTTPError for bad responses (4xx or 5xx)
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"An error occurred: {e}")
        return None
```

```python
# Call the search function with the defined search term
results_data = search_funders(search_term, url)

# Check if we got a valid result and then display it
if results_data and results_data.get('results'):
    # Convert the list of results into a pandas DataFrame
    df = pd.DataFrame(results_data['results'])
    
    # Select and reorder the columns for a cleaner display
    display_columns = ['id', 'display_name', 'country_code', 'works_count', 'cited_by_count']
    
    # Display the DataFrame. Jupyter Notebooks will automatically format this as a nice table.
    display(df[display_columns])
else:
    print(f"No results found for '{search_term}' or an error occurred.")
```
